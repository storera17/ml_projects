### *Background*

Professional sports continue to represent one of the most economically influential entertainment
sectors in the United States. The most commercially dominant organization continues to be the
National Football League (NFL). The league generates more than $18 billion annually [2], and 31
of 32 franchises reside within the 50 largest U.S. metropolitan areas [3]. This saturation severely
constrains domestic expansion potential, motivating strategic attention toward international
markets and underserved U.S. cities [4].

Global sports trends reinforce this shift. Recent international interest in American football is rising
[5], driving digital streaming, expanding broadcast rights, and enhancing fan-engagement
infrastructure [6–8]. Moreover, the NFL’s decision to schedule a record seven international games
during the 2025 season underscores the league’s commitment to international market cultivation
[9]. Thus, global and underrepresented domestic metros represent desirable markets to establish.
Keeping this in mind, a potential future expansion package establishing NFL franchises in these
candidate metros will require a creatively simple pitch. One way to address these parties in a
simple, underlying complex manner could be through the mapping of potential cities with the
current NFL franchises via a network approach. Unfortunately, starting with every city in the world
to find an exact solution for this network problem is computationally infeasible. A possible
solution is made viable through applying constraints on the set of all cities and populated regions
worldwide, V, filtered into a viable number of vertices belonging to the candidate set of cities 𝑉 !.
This was accomplished with a group of constraints coined ”filtering constraints” and explained
later.

The filtering and other constraints constitute a combination of demographic scale, economic
performance, commercial value potential, and feasible geographic placement features. Data from
the World Bank and Oxford Economics reveal that the world’s major metropolitan areas
disproportionately drive global GDP and long-run urban economic growth, which are critical
determinants for NFL franchise sustainability [10,11]. Additional considerations include
sponsorship markets, stadium financing, tourism, transportation infrastructure, and corporate
presence [12–17].

Herein, our team reports an approach to the NFL expansion problem by formulating a Binary
Integer Programming (BIP) model rooted in minimum dominating set theory in order to optimize
each set. Each candidate city is a node, and edges represent feasible fan-catchment or market-
coverage relationships. Unlike a classical unweighted dominating set, the NFL context introduces
demographic feasibility constraints, economic/financial viability constraints with net present value
(NPV) requirements, geographic separation and coverage constraints, stadium-cost and
operational-cost constraints, and expansion-budget limitations. Thus, a richer and more realistic
constrained minimum dominating set model is required. We prototype the model in Excel Solver
and implement the final scalable model in Python using Gurobi 11. Optimization results identify
the smallest feasible set of global cities that satisfy all market requirements and collectively
dominate coverage of the network.

### *Defining the Business Problem*

The NFL seeks to identify cities that are short-term, medium-term, and long-term candidates for new NFL franchises. Short-term candidates must satisfy all filtering constraints and are contained within the set of cities V<sub>3</sub>, medium-term candidates must satisfy at least two filtering constraints and are contained within the set of cities V<sub>2</sub>, and long-term candidates must satisfy at least one filtering constraint and are contained within the set of cities 𝑉<sub>1</sub>. Cities and populated regions that do not meet a single filtering constraint will not be considered as competitive candidate cities and are contained in the set 𝑉<sub>0</sub>. All cities and populated regions belong in the set of cities $V$. Additionally, all candidate cities (those within 𝑉<sub>3</sub>, 𝑉 <sub>2</sub>, 𝑎𝑛𝑑 𝑉 <sub>1</sub>) belong to the set of cities 𝑉<sub>c</sub>. Segmentations are described as:

* **Short-Term.** These cities are described by the colloquial saying the “best-of-the-best” targets. A city belonging to this segment must meet all filtering constraints. As a result, these cities belong to a group of cities that make up the set V<sub>3</sub>.
* **Medium-Term.** Each city belonging to this segment requires satisfying at least two of the three filtering constraints. When a city belongs to this segment, it is considered a moderate target candidate city. Cities in the medium-term segment belong to the set of cities V<sub>2</sub>.
* **Long-Term.** These cities meet at least one of the filtering constraints. These cities are poor candidates and belong to the set of cities in V<sub>1</sub>.
* **Non-Candidate Cities.** These represent the set of cities in $V$ that are not selected and meet no filtering requirement.

The aforementioned approach to solving the placement of an NFL franchise problem discussed using a BIP model rooted in minimum dominating set theory. The initial set of cities and populated areas in $V$ is too large for a computationally feasible exact solution to be determined. Thus, there is a requirement for the aforementioned set of constraints to collapse $V$ into a set of viable cities V<sub>c</sub>. The cities that operate the best minimum dominating set of target candidates for a new NFL
franchise would belong to V<sub>3</sub>. In all, a city belongs to the set 𝑉<sub>3</sub> if it satisfies the population and feasibility constraints, economic and financial constraints with net present value (NPV) thresholds, geographic adjacency and market coverage constraints, and budget constraints. To summarize, the NFL’s objective is to select the smallest number of cities that jointly provide complete coverage of the network while satisfying all feasibility constraints. This structure is equivalent to a constrained minimum dominating set problem with feasibility-restricted decision variables.

### *Data Acquisition*

#### Population & Demographic Features

__________

The construction of the optimization model requires a comprehensive dataset describing
demographic, economic, financial, and geographic characteristics of a large set of global metropolitan regions. As discussed in Section 1, the initial universe of cities $V$ is far too large to admit a tractable optimization model. Thus, the purpose of the data acqusition process is twofold. First, to assemble reliable and comparable data describing the relevant attributes of each city. Secondly, to derive the filtering constraints that reduce the universe $V$ to the viable set of candidate
cities V<sub>c</sub>, from which the short-term feasible set V<sub>3</sub> is identified. The approach follows a structure similar to other large-scale screening methodologies in applied operations research, wherein a combination of demographic suitability, economic viability, and geographic adjacency information is used to prune the decision space before formal optimization. Population and demographic data constitute the first major category of filtering criteria. These data capture the fundamental requirement that a metropolitan region must possess a sufficiently large
and stable population base to support an NFL franchise. For U.S. cities, we employ population estimates, density measures, household characteristics, age structure, and socioeconomic indicators obtained from the U.S. Census Bureau’s metropolitan and micropolitan statistical tables and Gazetteer geographic files [3,22]. International population and demographic attributes are drawn from the United Nations World Urbanization Prospects and from the World Bank’s World Development Indicators, which provide harmonized estimates of metropolitan population aggregates for global urban agglomerations [23,24]. Additional demographic features, including
birth and death rates, migration trends, household mobility, and regional age distributions, were obtained from OECD demographic indicators and CDC vital statistics [24–26]. Because metropolitan population alone is often an insufficient proxy for market size in the professional sports context, we further incorporate tourism volumes from UNWTO statistics, safety indices from Numbeo, and crime measures from UNODC [26–30], allowing the filtering mechanism to differentiate between markets of comparable size but disparate stability, safety, or visitor exposure.

#### Economic & Value Metrics

______

Economic and value-related data form the second major category of filtering constraints and support the requirement that each candidate market must demonstrate financial capacity to sustain long-run franchise operations. Metropolitan GDP, GDP per capita, and projected economic growth rates are obtained from the World Bank and from Oxford Economics’ Global Cities reports [10,11,29]. These indicators approximate the underlying economic strength of the region and act as proxies for future revenue potential. Measures of labor force participation, wage levels, employment conditions, and cost-of-living adjustments are drawn from U.S. Bureau of Labor Statistics data and from OECD economic datasets [31]. Corporate presence and business intensity, including counts of Fortune 500 headquarters, broader headquarters concentrations, and startup density, are derived from Fortune, Orbis, Crunchbase, and PitchBook repositories, all of which offer detailed, location-linked microeconomic information [32–35]. Innovation, intellectual property activity, and regional research expenditures are quantified using WIPO patent statistics and NSF research expenditure data [36,37]

Financial feasibility also depends on cost-side considerations. Stadium construction cost estimates and infrastructure expenses are derived from RSMeans construction indices and KPMG stadium-financing assessments [12,39]. Tax burden, municipal fiscal health, and general cost-of-doing-business metrics use data from the Tax Foundation and other governmental sources [38]. Together, these economic indicators allow us to compute whether each city satisfies minimum revenue and profitability requirements and to approximate the cost component 𝑐& used later in the optimization model.

#### Geographic & Network Features
________

The third major data component concerns geographic and network-structure information. Because NFL fan bases exhibit strong spatial dependence and because the optimization model relies on a
dominating-set formulation, it is essential to determine whether one metropolitan area can realistically “cover” another. Accordingly, geographic coordinates for all cities are obtained from the U.S. Census Gazetteer and the GeoNames global cities database [22,44]. Distances between all city pairs are computed using the great-circle (Haversine) formula, following the standard approach described by Sinnott (1984) [50]. These distances determine the adjacency parameters $a$<sub>ij</sub>, where $𝑎$<sub>ij</sub> = 1 if city $i$ lies within the prescribed catchment radius of city $j$. Additional transportation and accessibility attributes—including airport throughput, global connectivity scores, and travel-time approximations are drawn from IATA statistics, international transport datasets, and routing services such as OSRM and Google Distance Matrix [45–47]. Mobility and congestion data from the TomTom Index provide further refinement of regional accessibility characteristics [49], which can alter effective catchment boundaries and thereby influence adjacency structure.

Collectively, these data sources enable the derivation of the filtering constraints that reduce the initial set $V$ to the viable set $V$<sub>c</sub>. Cities that satisfy all relevant demographic and economic feasibility criteria are assigned to $V$<sub>3</sub>, the set of markets eligible for immediate expansion. Cities meeting only a subset of constraints populate intermediary segments $V$<sub>1</sub> and $V$<sub>1</sub>, which may carry strategic value in medium-term or long-term planning scenarios. The resulting structured dataset supports the construction of the geographic network on which the optimization model is defined and provides the parameters necessary for enforcing feasibility, coverage, and budget constraints within the binary programming formulation presented in Section 3.

</div>

The optimization model developed for the NFL expansion problem builds upon the dominating-set framework described earlier and integrates demographic, economic, and financial feasibility requirements into a coherent binary integer programming formulation. Let $G = (V, E)$ denote the network constructed from the full set of global metropolitan areas, where edges encode coverage relationships based on great-circle distances. As established in Section 2, the filtering constraints determine the viable set of candidate cities $V_C \subseteq V$, while the subset $V_S \subseteq V_C$ includes only those cities that satisfy all feasibility requirements and are thus eligible for immediate franchise placement. 

For each feasible city $i \in V_S$, a binary variable $x_i$ indicates whether a franchise is placed in city $i$. The feasibility of a city is governed by two precomputed indicators: a population suitability measure $\phi_i$ and an economic/value viability measure $\theta_i$. These indicators restrict selection to cities that simultaneously satisfy all demographic and financial requirements.

Market coverage is represented through adjacency parameters $a_{ij}$, where $a_{ij} = 1$ when city $i$ is capable of covering city $j$ within the prescribed catchment radius, and $a_{ij} = 0$ otherwise. This structure enables the model to enforce the classical dominating-set requirement that each city in $V_C$ must either host a franchise itself or lie within the coverage region of one. The complete linear mathematical model is broken down into subsequent pieces to explain each component, then presented altogether. Beginning with the objective function, the goal is to minimize the number of selected franchise locations as:

$$
\min \sum_{i \in V_S} x_i
\tag{1a}
$$ 

which will be interesting to see where cities would be selected with respect to the current NFL landscape. The feasibility constraints come next. These show that a city may be selected only if it satisfies the population and value constraints encoded by $\phi_i$ and $\theta_i$:

$$
x_i \leq \phi_i, \quad i \in V_S
\tag{1c}
$$

$$
x_i \leq \theta_i, \quad i \in V_S
\tag{1d}
$$

If either indicator equals zero, constraints (1c)-(1d) force $x_i = 0$, thereby eliminating the city from consideration. For the budget constraint, let $c_i$ denote the total capital cost associated with establishing a franchise in city $i$. The aggregate cost of selected cities must not exceed the expansion budget $B$:

$$
\sum_{i \in V_S} c_i x_i \leq B
\tag{1e}
$$

The market separation constraints specify that certain pairs of cities $(i,k)$ may not simultaneously host franchises due to geographic proximity or market cannibalization considerations. Let $S$ denote the set of such pairs. For each $(i,k) \in S$, the model imposes:

$$
x_i + x_k \leq 1, \quad (i,k) \in S
\tag{1f}
$$

The number-of-teams constraint is next. If the league imposes a cap $n_{\max}$ on the number of new franchises, the model includes:

$$
\sum_{i \in V_S} x_i \leq n_{\max}
\tag{1g}
$$

The integrality constraint requires all decision variables to be binary:

$$
x_i \in \{0,1\}, \quad i \in V_S
\tag{1h}
$$

In addition to these structural constraints, we formally define the filtering indicators that determine feasibility prior to optimization. To complete the mathematical framework, we describe the construction of the population and value feasibility indicators $\phi_i$ and $\theta_i$, which are used in constraints (1c)-(1d). Let $\Phi$ denote the set of population-based and demographic-based criteria. For each city $i$ and each population sub-criterion $k \in \Phi$, define the binary variable:

$$
y_{ik} =
\begin{cases}
1, & \text{if city } i \text{ satisfies population criterion } k \\
0, & \text{otherwise}
\end{cases}
\tag{1i}
$$

Similarly, let $\Theta$ denote the set of economic and financial sub-constraints. For each city $i$ and each economic/value sub-criterion $m \in \Theta$, define:

$$
z_{im} =
\begin{cases}
1, & \text{if city } i \text{ satisfies economic/value criterion } m \\
0, & \text{otherwise}
\end{cases}
\tag{1j}
$$

Population feasibility requires meeting a minimum number of demographic criteria, represented by the threshold $\tau_{\phi}$:

$$
\phi_i =
\begin{cases}
1, & \text{if } \sum_{k \in \Phi} y_{ik} \geq \tau_{\phi} \\
0, & \text{otherwise}
\end{cases}
\tag{1k}
$$

Similarly, economic feasibility requires satisfying at least $\tau_{\theta}$ financial criteria:

$$
\theta_i =
\begin{cases}
1, & \text{if } \sum_{m \in \Theta} z_{im} \geq \tau_{\theta} \\
0, & \text{otherwise}
\end{cases}
\tag{1l}
$$

A central economic requirement concerns long-run financial viability, characterized by projected net present value. Let $NPV_i$ denote the projected net present value of future cash flows associated with market $i$, incorporating revenues, operational costs, startup expenditures, and discounting. A city must meet the minimum financial return requirement:

$$
NPV_i \geq NPV_{\min}
\tag{1m}
$$

with a dedicated sub-indicator:

$$
z_{i,NPV} =
\begin{cases}
1, & \text{if } NPV_i \geq NPV_{\min} \\
0, & \text{otherwise}
\end{cases}
\tag{1n}
$$

These filtering constraints collectively determine membership in $V_C$ and $V_S$, ensuring that infeasible cities are eliminated prior to optimization and that all selected cities satisfy the minimum demographic and financial requirements necessary for sustainable league expansion. The complete formulation of the NFL expansion problem is given below:

$$
\min \sum_{i \in V_S} x_i
\tag{1a}
$$

subject to:

$$
\sum_{i \in V_S} a_{ij}x_i \geq 1, \quad \forall j \in V_C
\tag{1b}
$$

$$
x_i \leq \phi_i, \quad \forall i \in V_S
\tag{1c}
$$

$$
x_i \leq \theta_i, \quad \forall i \in V_S
\tag{1d}
$$

$$
\sum_{i \in V_S} c_i x_i \leq B
\tag{1e}
$$

$$
x_i + x_k \leq 1, \quad \forall (i,k) \in S
\tag{1f}
$$

$$
\sum_{i \in V_S} x_i \leq n_{\max}
\tag{1g}
$$

$$
x_i \in \{0,1\}, \quad \forall i \in V_S
\tag{1h}
$$

$$
y_{ik} =
\begin{cases}
1, & \text{if city } i \text{ satisfies population criterion } k \\
0, & \text{otherwise}
\end{cases}
\tag{1i}
$$

$$
z_{im} =
\begin{cases}
1, & \text{if city } i \text{ satisfies economic/value criterion } m \\
0, & \text{otherwise}
\end{cases}
\tag{1j}
$$

$$
\phi_i =
\begin{cases}
1, & \text{if } \sum_{k \in \Phi} y_{ik} \geq \tau_{\phi} \\
0, & \text{otherwise}
\end{cases}
\tag{1k}
$$

$$
\theta_i =
\begin{cases}
1, & \text{if } \sum_{m \in \Theta} z_{im} \geq \tau_{\theta} \\
0, & \text{otherwise}
\end{cases}
\tag{1l}
$$

$$
NPV_i \geq NPV_{\min}
\tag{1m}
$$

$$
z_{i,NPV} =
\begin{cases}
1, & \text{if } NPV_i \geq NPV_{\min} \\
0, & \text{otherwise}
\end{cases}
\tag{1n}
$$

### *Data Pipeline & Preprocessing*

The computational procedure begins with the integrated dataset described in Section 1. Each row of the master table corresponds to a metropolitan market and contains:

- a unique city identifier and printable city name,
- geographic coordinates (latitude, longitude),
- stadium and startup cost estimates $c_i$,
- projected net present value $NPV_i$,
- four population/demographic sub-criteria $\{y_{ik}\}_{k=1}^{4}$,
- four economic/value sub-criteria $\{z_{im}\}_{m=1}^{4}$,
- precomputed indicators describing geographic accessibility and catchment status.

The first preprocessing stage computes the overarching population and economic feasibility indicators $\phi_i$ and $\theta_i$ in direct accordance with Equations (1k)-(1l). For each city, the code counts how many of the 4 population criteria $y_{ik}$ are satisfied and sets

$$
\phi_i =
\begin{cases}
1, & \text{if } \sum_{k \in \Phi} y_{ik} \geq \tau_{\phi}, \\
0, & \text{otherwise.}
\end{cases}
\tag{1k}
$$

with an analogous rule for $\theta_i$ based on the economic/value criteria $z_{im}$ and threshold $\tau_{\theta}$. In the baseline experiments, both thresholds are set to seven out of ten, which enforces a relatively strict notion of feasibility while still leaving a meaningful number of candidate markets. Cities failing either $\phi_i = 1$ or $\theta_i = 1$ are removed from the feasible set $V_S$ prior to optimization, while cities that satisfy at least one filtering category are retained in the broader candidate set $V_C$.

The next preprocessing stage constructs the geographic network. Using each city’s latitude and longitude, the code evaluates pairwise great-circle distances via the Haversine formula [50]. For a given coverage radius $R$ (150 miles in the baseline scenario), the adjacency indicator is defined programmatically as

$$
a_{ij} =
\begin{cases}
1, & \text{if distance}(i,j) \leq R, \\
0, & \text{otherwise.}
\end{cases}
$$

This adjacency matrix serves as the edge set of $G = (V, E)$ and directly enters the coverage constraints (1b). The same distance engine is later used to identify “too close” market pairs $(i,k) \in S$ for which two franchises would be considered cannibalizing, and thus must satisfy the separation constraint (1f).

Finally, the code constructs segmentation labels for each market (belonging to $V_N$, $V_L$, $V_M$, or $V_S$) based on how many of the population, economic, and geographic filters it satisfies. This segmentation is stored with the data so that the optimization can either be restricted to the most viable set $V_S$ or extended to include medium-term and long-term markets in alternative scenarios.

### *Gurobi Model Construction*

Once the data pipeline is complete, the Python notebook instantiates a Gurobi model named *NFL_Expansion*. The implementation follows the exact structure of the mathematical program (1a)-(1n).

**Decision variables.** For each fully feasible city $i \in V_S$, the code creates a binary variable

$$
x_i =
\begin{cases}
1, & \text{if a new or relocated NFL franchise is placed in city } i, \\
0, & \text{otherwise.}
\end{cases}
$$

All variables are declared simultaneously as a dictionary-like structure, which simplifies looping and allows Gurobi to exploit sparsity in the constraint matrix.

**Objective function.** The objective is to minimize the number of selected franchises, consistent with Equation (1a):

$$
\min \sum_{i \in V_S} x_i
$$

In code, this is expressed as a single linear expression constructed with Gurobi’s `quicksum` operator. This choice directly encodes the minimum dominating set philosophy: the league wishes to cover as many markets as possible using the smallest feasible number of teams.

### *Constraint Implementation*

#### Coverage Constraint (1b)

_____

For each market $j \in V_C$, the code adds

$$
\sum_{i \in V_S} a_{ij}x_i \geq 1
$$

which ensures that every viable market is either a host city or lies within the catchment region of at least one selected franchise. The adjacency coefficients $a_{ij}$ are read from the precomputed distance matrix and only nonzero terms are included, which keeps the model sparse.

#### Feasibility Constraints (1c)-(1d)
____

The filters are enforced by tying each decision variable to its corresponding feasibility indicators

$$
x_i \leq \phi_i, \quad x_i \leq \theta_i, \quad i \in V_S.
$$

If either indicator equals zero, the city cannot be selected, and the integrality of $x_i$ guarantees that infeasible locations are eliminated from consideration.

#### Budget Constraint (1e)
___

The total up-front stadium and startup cost of the chosen expansion package must respect a league-wide budget $B$:

$$
\sum_{i \in V_S} c_i x_i \leq B.
$$

The code reads $c_i$ from the financial dataset and constructs a single linear inequality. In the baseline solution displayed later, the budget was set to approximately \$337.1 billion, which is sufficiently large to allow coverage of all 174 markets while still binding at optimality.

#### Separation Constraint (1f)
___

For each pair of markets $(i,k) \in S$ that are judged too close to sustain independent franchises, the model adds

$$
x_i + x_k \leq 1.
$$

The set $S$ is derived from the same distance matrix used for adjacency, but with a much smaller separation radius (e.g., a few tens of miles) and can be expanded to reflect qualitative judgments about overlapping media markets.

#### Team-Cap Constraint (1g)
____

When the league specifies a hard cap $n_{\max}$ on the total number of new teams, the model enforces

$$
\sum_{i \in V_S} x_i \leq n_{\max}.
$$

In some experiments this constraint is relaxed to let the budget and coverage conditions drive the solution; in others it is used to simulate expansions of specific sizes, such as four new franchises.

Integrality conditions (1h) are satisfied by declaring all decision variables as binary at model creation. Gurobi automatically handles these constraints during branch-and-bound search.

### *Feasibility & Coverage Properties*

The first set of experiments was designed to verify that the model behaved as intended and that the dominating set structure was correctly enforced. Under the baseline configuration, where both the population and economic filters were active, the coverage radius was fixed at 150 miles, and the full set of 174 viable markets was included, the optimization produced a solution that satisfied all constraints. Every market in the candidate set $V_C$ satisfied the condition

$$
\sum_{i \in V_S} a_{ij}x_i \geq 1.
$$

Each market was therefore selected as a franchise site or fell within the catchment region of at least one selected city. No city that failed to meet the population or economic filter conditions was chosen. This confirmed that the filter structure was implemented correctly. The capital expenditure constraint was also exactly tight, which showed that the budget was a binding and important part of the problem. The baseline solution contained 71 selected cities. This relatively large dominating set reflects the sparse adjacency structure implied by a 150-mile radius. Many markets have no neighbors within this distance, which forces the model to open many facilities to maintain global coverage.

### *Geographic Structure of the Baseline Solution*

The geographic pattern of the selected 71 markets reveals that spatial structure, not economic scale, is the dominant driver of the solution. The model often did not select major world markets such as New York, Los Angeles, London, Tokyo, or Shanghai. Instead, it frequently selected medium sized or regionally central cities that were better positioned to dominate nearby clusters within the fixed coverage radius. In regions with dense populations, including the eastern United States, Western Europe, and Southeast Asia, many markets lie outside the 150-mile radius of any single neighboring city. To maintain feasibility, the solver was required to open several facilities in relatively close geographic proximity. The same pattern appears in inland areas where population centers are widely separated. The solver had to select geographically isolated cities to prevent uncovered gaps in the network. Although the selected cities are not always the largest in terms of demographic or economic potential, the resulting pattern still forms a coherent worldwide coverage system. Every major geographic area remains represented, but the selections reflect the structure of the coverage graph rather than strategic preferences of the league.

### *Sensitivity to Filter Stringency*

A sensitivity analysis over the filter thresholds $\tau_{\phi}$ and $\tau_{\theta}$ was conducted to assess the stability of the model. The thresholds were varied across the set $\{1,2,3,4\}$. The results show a sharp discontinuity in feasibility. The model was feasible only for the four combinations $(1,1)$, $(1,2)$, $(2,1)$, and $(2,2)$. In all four cases, the solution selected exactly 71 markets, the total cost remained essentially unchanged, and the set of selected cities changed only slightly. This means that the feasible region is very narrow but internally stable. Within this region, the solution is largely determined by adjacency and the budget rather than by the specific filter thresholds.

When either threshold is increased to three or four, feasibility fails immediately. The filter removes too many markets from the eligible set, and the remaining candidates cannot dominate the full network under the fixed radius. The solver correctly reports that coverage is impossible. This shows that the filters do not create a gradual trade off between selectivity and coverage. Instead, there is a small region in which the model behaves consistently, followed by a hard boundary where the problem no longer has a solution.

### *Sensitivity to Budget & Coverage Radius*

Although the experiments did not vary the budget or the radius, their influence can be understood from the structure of the model. Because the objective is to minimize the number of facilities, increasing the budget does not cause the solver to open extra cities. Decreasing the budget, however, restricts the model’s ability to choose certain locations. If the budget is too small, there is no selection of sites that both satisfies the cost constraint and dominates the full network, and the model becomes infeasible. The radius is the dominant geometric factor in the size of the dominating set. A larger radius increases adjacency among cities and allows the model to cover many more markets with fewer facilities. A smaller radius reduces adjacency, increases the number of facilities required, and narrows the feasible region of filter thresholds. Although the radius was fixed in these experiments, this parameter is the primary structural determinant of the scale and feasibility of global expansion in this model.

### *Key Insights*

The results show that global NFL expansion is fundamentally a network design problem. The dominating set formulation clarifies that the league must select a group of cities that collectively cover all viable markets. The optimal network often uses cities that are geographically strategic rather than economically dominant. The population and economic filters play a major role by defining the feasible set before the solver is called. The model shows that meaningful trade offs exist between coverage, cost, and market quality, and that these trade offs become especially clear when examining the filter thresholds or the coverage radius. The framework therefore serves as an effective tool for exploring different strategic priorities in expansion planning.

### *Limitations*

The model uses fixed point estimates for demographics, NPV, fan interest, and other parameters. These values are uncertain in practice, and the model does not account for this uncertainty. The formulation also assumes a single period decision process. Real expansion unfolds over many decades and involves sequential decision making and feedback effects. The representation of competitive effects is limited. The model does not capture the effect of competition with existing NFL teams or other sports leagues, nor does it represent political or regulatory constraints that often influence stadium development.

### *Possible Extensions*

Future enhancements could include multi objective formulations that incorporate financial performance or fan engagement into the objective function. Stochastic or robust optimization approaches could be used to incorporate uncertainty in economic or demographic projections. Another possible extension is to integrate the dominating set model with scheduling or travel considerations, which would allow expansion planning to account for travel distances, time zone effects, and broadcast logistics.
