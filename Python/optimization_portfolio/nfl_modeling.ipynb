### *Background*

Professional sports continue to represent one of the most economically influential entertainment
sectors in the United States. The most commercially dominant organization continues to be the
National Football League (NFL). The league generates more than $18 billion annually [2], and 31
of 32 franchises reside within the 50 largest U.S. metropolitan areas [3]. This saturation severely
constrains domestic expansion potential, motivating strategic attention toward international
markets and underserved U.S. cities [4].

Global sports trends reinforce this shift. Recent international interest in American football is rising
[5], driving digital streaming, expanding broadcast rights, and enhancing fan-engagement
infrastructure [6–8]. Moreover, the NFL’s decision to schedule a record seven international games
during the 2025 season underscores the league’s commitment to international market cultivation
[9]. Thus, global and underrepresented domestic metros represent desirable markets to establish.
Keeping this in mind, a potential future expansion package establishing NFL franchises in these
candidate metros will require a creatively simple pitch. One way to address these parties in a
simple, underlying complex manner could be through the mapping of potential cities with the
current NFL franchises via a network approach. Unfortunately, starting with every city in the world
to find an exact solution for this network problem is computationally infeasible. A possible
solution is made viable through applying constraints on the set of all cities and populated regions
worldwide, V, filtered into a viable number of vertices belonging to the candidate set of cities 𝑉 !.
This was accomplished with a group of constraints coined ”filtering constraints” and explained
later.

The filtering and other constraints constitute a combination of demographic scale, economic
performance, commercial value potential, and feasible geographic placement features. Data from
the World Bank and Oxford Economics reveal that the world’s major metropolitan areas
disproportionately drive global GDP and long-run urban economic growth, which are critical
determinants for NFL franchise sustainability [10,11]. Additional considerations include
sponsorship markets, stadium financing, tourism, transportation infrastructure, and corporate
presence [12–17].

Herein, our team reports an approach to the NFL expansion problem by formulating a Binary
Integer Programming (BIP) model rooted in minimum dominating set theory in order to optimize
each set. Each candidate city is a node, and edges represent feasible fan-catchment or market-
coverage relationships. Unlike a classical unweighted dominating set, the NFL context introduces
demographic feasibility constraints, economic/financial viability constraints with net present value
(NPV) requirements, geographic separation and coverage constraints, stadium-cost and
operational-cost constraints, and expansion-budget limitations. Thus, a richer and more realistic
constrained minimum dominating set model is required. We prototype the model in Excel Solver
and implement the final scalable model in Python using Gurobi 11. Optimization results identify
the smallest feasible set of global cities that satisfy all market requirements and collectively
dominate coverage of the network.

### *Defining the Business Problem*

The NFL seeks to identify cities that are short-term, medium-term, and long-term candidates for new NFL franchises. Short-term candidates must satisfy all filtering constraints and are contained within the set of cities V<sub>3</sub>, medium-term candidates must satisfy at least two filtering constraints and are contained within the set of cities V<sub>2</sub>, and long-term candidates must satisfy at least one filtering constraint and are contained within the set of cities 𝑉<sub>1</sub>. Cities and populated regions that do not meet a single filtering constraint will not be considered as competitive candidate cities and are contained in the set 𝑉<sub>0</sub>. All cities and populated regions belong in the set of cities $V$. Additionally, all candidate cities (those within 𝑉<sub>3</sub>, 𝑉 <sub>2</sub>, 𝑎𝑛𝑑 𝑉 <sub>1</sub>) belong to the set of cities 𝑉<sub>c</sub>. Segmentations are described as:

* **Short-Term.** These cities are described by the colloquial saying the “best-of-the-best” targets. A city belonging to this segment must meet all filtering constraints. As a result, these cities belong to a group of cities that make up the set V<sub>3</sub>.
* **Medium-Term.** Each city belonging to this segment requires satisfying at least two of the three filtering constraints. When a city belongs to this segment, it is considered a moderate target candidate city. Cities in the medium-term segment belong to the set of cities V<sub>2</sub>.
* **Long-Term.** These cities meet at least one of the filtering constraints. These cities are poor candidates and belong to the set of cities in V<sub>1</sub>.
* **Non-Candidate Cities.** These represent the set of cities in $V$ that are not selected and meet no filtering requirement.

The aforementioned approach to solving the placement of an NFL franchise problem discussed using a BIP model rooted in minimum dominating set theory. The initial set of cities and populated areas in $V$ is too large for a computationally feasible exact solution to be determined. Thus, there is a requirement for the aforementioned set of constraints to collapse $V$ into a set of viable cities V<sub>c</sub>. The cities that operate the best minimum dominating set of target candidates for a new NFL
franchise would belong to V<sub>3</sub>. In all, a city belongs to the set 𝑉<sub>3</sub> if it satisfies the population and feasibility constraints, economic and financial constraints with net present value (NPV) thresholds, geographic adjacency and market coverage constraints, and budget constraints. To summarize, the NFL’s objective is to select the smallest number of cities that jointly provide complete coverage of the network while satisfying all feasibility constraints. This structure is equivalent to a constrained minimum dominating set problem with feasibility-restricted decision variables.

### *Data Acquisition*

#### Population & Demographic Features

__________

The construction of the optimization model requires a comprehensive dataset describing
demographic, economic, financial, and geographic characteristics of a large set of global metropolitan regions. As discussed in Section 1, the initial universe of cities $V$ is far too large to admit a tractable optimization model. Thus, the purpose of the data acqusition process is twofold. First, to assemble reliable and comparable data describing the relevant attributes of each city. Secondly, to derive the filtering constraints that reduce the universe $V$ to the viable set of candidate
cities V<sub>c</sub>, from which the short-term feasible set V<sub>3</sub> is identified. The approach follows a structure similar to other large-scale screening methodologies in applied operations research, wherein a combination of demographic suitability, economic viability, and geographic adjacency information is used to prune the decision space before formal optimization. Population and demographic data constitute the first major category of filtering criteria. These data capture the fundamental requirement that a metropolitan region must possess a sufficiently large
and stable population base to support an NFL franchise. For U.S. cities, we employ population estimates, density measures, household characteristics, age structure, and socioeconomic indicators obtained from the U.S. Census Bureau’s metropolitan and micropolitan statistical tables and Gazetteer geographic files [3,22]. International population and demographic attributes are drawn from the United Nations World Urbanization Prospects and from the World Bank’s World Development Indicators, which provide harmonized estimates of metropolitan population aggregates for global urban agglomerations [23,24]. Additional demographic features, including
birth and death rates, migration trends, household mobility, and regional age distributions, were obtained from OECD demographic indicators and CDC vital statistics [24–26]. Because metropolitan population alone is often an insufficient proxy for market size in the professional sports context, we further incorporate tourism volumes from UNWTO statistics, safety indices from Numbeo, and crime measures from UNODC [26–30], allowing the filtering mechanism to differentiate between markets of comparable size but disparate stability, safety, or visitor exposure.

#### Economic & Value Metrics

______

Economic and value-related data form the second major category of filtering constraints and support the requirement that each candidate market must demonstrate financial capacity to sustain long-run franchise operations. Metropolitan GDP, GDP per capita, and projected economic growth rates are obtained from the World Bank and from Oxford Economics’ Global Cities reports [10,11,29]. These indicators approximate the underlying economic strength of the region and act as proxies for future revenue potential. Measures of labor force participation, wage levels, employment conditions, and cost-of-living adjustments are drawn from U.S. Bureau of Labor Statistics data and from OECD economic datasets [31]. Corporate presence and business intensity, including counts of Fortune 500 headquarters, broader headquarters concentrations, and startup density, are derived from Fortune, Orbis, Crunchbase, and PitchBook repositories, all of which offer detailed, location-linked microeconomic information [32–35]. Innovation, intellectual property activity, and regional research expenditures are quantified using WIPO patent statistics and NSF research expenditure data [36,37]

Financial feasibility also depends on cost-side considerations. Stadium construction cost estimates and infrastructure expenses are derived from RSMeans construction indices and KPMG stadium-financing assessments [12,39]. Tax burden, municipal fiscal health, and general cost-of-doing-business metrics use data from the Tax Foundation and other governmental sources [38]. Together, these economic indicators allow us to compute whether each city satisfies minimum revenue and profitability requirements and to approximate the cost component 𝑐& used later in the optimization model.
